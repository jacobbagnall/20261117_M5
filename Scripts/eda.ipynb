```python
#import and view data
import pandas

df_book = pandas.read_csv("data/03_Library Systembook.csv")

df_system = pandas.read_csv("data/03_Library Systemcustomers.csv")

print(df_book.head())
print(df_system.head())

#print(df_book)
#print(df_system)

#print(df_book.shape)
#print(df_system.shape)

#print(df_book.describe(include="all"))
#print(df_system.describe(include="all"))
```

```python
#data clean
import pandas
# strips ""
df_book["Book checkout"] = df_book["Book checkout"].str.strip().str.strip("\"'")

# set colum data types
df_book["Id"] = pandas.to_numeric(df_book["Id"], errors = "coerce").astype("Int64")
df_book["Books"] = df_book["Books"].astype("string")
df_book["Book checkout"] = pandas.to_datetime(df_book["Book checkout"], format= "%d/%m/%Y", errors = "coerce")
df_book["Book Returned"] = pandas.to_datetime(df_book["Book Returned"], format= "%d/%m/%Y", errors = "coerce")
df_book["Days allowed to borrow"] = df_book["Days allowed to borrow"].astype("string")
df_book["Customer ID"] = pandas.to_numeric(df_book["Customer ID"], errors = "coerce").astype("Int64")

#remove null rows and save in failed variable
required = ["Id","Books","Book checkout","Book Returned", "Days allowed to borrow", "Customer ID"
 ]
fail_mask = df_book[required].isna().any(axis=1)

failed_book_df = df_book[fail_mask].copy()
failed_book_df["fail_reason"] = "NULL Value"
df_book_no_null = df_book[~fail_mask].copy()

#duplicate PK ID
id_mask = df_book_no_null["Id"].duplicated(keep=False)

id_failed = df_book_no_null[id_mask].copy()
id_failed["fail_reason"] = "duplicate_id"

failed_book_df = pandas.concat([failed_book_df,id_failed] )

books_cleanned = df_book_no_null[~id_mask.copy()]


#df_system_clean

# set colum data types
df_system["Customer ID"] = pandas.to_numeric(df_system["Customer ID"], errors = "coerce").astype("Int64")
df_system["Customer Name"] = df_system["Customer Name"].astype("string")

#remove null rows and save in failed variable
required = ["Customer ID","Customer Name"]
fail_mask = df_system[required].isna().any(axis=1)

failed_system_df = df_system[fail_mask].copy()
failed_system_df["fail_reason"] = "NULL Value"
df_system_no_null = df_system[~fail_mask].copy()

#duplicate PK ID
id_mask = df_system_no_null["Customer ID"].duplicated(keep=False)

id_failed = df_system_no_null[id_mask].copy()
id_failed["fail_reason"] = "duplicate_id"

failed_system_df = pandas.concat([failed_system_df,id_failed] )

system_cleanned = df_system_no_null[~id_mask.copy()]


```

```python
# export to csv in cleanned

books_cleanned.to_csv ("cleanned/books_cleanned.csv", index=False, date_format="%d/%m/%Y")
system_cleanned.to_csv ("cleanned/system_cleanned.csv", index=False, date_format="%d/%m/%Y")
failed_book_df.to_csv ("cleanned/failed_book_df.csv", index=False, date_format="%d/%m/%Y")
failed_system_df.to_csv ("cleanned/failed_system_df.csv", index=False, date_format="%d/%m/%Y")

print ("cleaned exported")
```
